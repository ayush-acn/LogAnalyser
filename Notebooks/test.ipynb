```python
import boto3

bedrock = boto3.client(
    "bedrock",
    region_name="us-east-1"
)

response = bedrock.list_inference_profiles()

print("=" * 80)

for profile in response["inferenceProfileSummaries"]:
    print("Name :", profile["inferenceProfileName"])
    print("ID   :", profile["inferenceProfileId"])
    print("ARN  :", profile["inferenceProfileArn"])
    print("Status:", profile["status"])
    print("-" * 80)
```

```python
import boto3
import json

client = boto3.client(
    "bedrock-runtime",
    region_name="us-east-1"
)

body = {
    "anthropic_version": "bedrock-2023-05-31",
    "max_tokens": 200,
    "messages": [
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "Tell me a joke."
                }
            ]
        }
    ]
}

response = client.invoke_model(
    modelId="us.anthropic.claude-sonnet-4-6",
    body=json.dumps(body),
    contentType="application/json",
    accept="application/json"
)

response_body = json.loads(response["body"].read())

print(response_body["content"][0]["text"])
```

```python
from langchain_aws import ChatBedrockConverse

llm = ChatBedrockConverse(
    model="us.anthropic.claude-sonnet-4-6",
    region_name="us-east-1",
    temperature=0,
    max_tokens=200,
)

response = llm.invoke("Tell me a joke about software engineers.")

print(response.content)
```

```python
from langchain_aws import ChatBedrockConverse
from langchain_core.messages import HumanMessage

llm = ChatBedrockConverse(
    model="us.anthropic.claude-sonnet-4-6",
    region_name="us-east-1",
)

response = llm.invoke([
    HumanMessage(content="Explain recursion in simple terms.")
])

print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_aws import ChatBedrockConverse

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert Python tutor."),
    ("human", "{question}")
])

llm = ChatBedrockConverse(
    model="us.anthropic.claude-sonnet-4-6",
    region_name="us-east-1",
)

chain = prompt | llm

response = chain.invoke({
    "question": "What is a Python decorator?"
})

print(response.content)
```
